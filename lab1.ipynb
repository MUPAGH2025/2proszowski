```python
import pint as pnt
from scipy import constants as const
from matplotlib import pyplot as plt
import numpy as np
from open_atmos_jupyter_utils import show_plot 
si = pnt.UnitRegistry()
```

```python
l = 100*si.m + 50*si.cm + 200*si.mm
l = l/10/si.ms**2
l.to_base_units()
```

```python
def rho(p, T, Rd):
    return p / (Rd * T)
```

```python
R = 8.314 * si.J / (si.kg * si.K)
massMolarDryAir =  28.966 * si.kg / si.kmol

Rd = const.R / massMolarDryAir* si.J / (si.kg * si.K)
print(Rd.to_base_units() * si.J / (si.kg * si.K))
print(R/massMolarDryAir.to_base_units())
rho(1013*si.hPa, 296*si.K, Rd).to_base_units()
```

```python
si.setup_matplotlib()
```

```python
plt.plot(
        T := np.linspace(280, 320, 100) * si.K , 
         rho(p=1013*si.hPa, T=T, Rd=Rd).to_base_units(),
         label="Gestosc powietrza w sali"
         )
plt.xlabel("Temperature [°C]")
plt.ylabel("Density [kg/m^3]")
plt.legend()
plt.gca().xaxis.set_units(si.celsius)
plt.grid()
show_plot("fig1_gestosc")
```
